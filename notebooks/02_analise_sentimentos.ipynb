# **Análise de Sentimentos**

## Importação das bibliotecas e leitura dos dados

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
avaliacoes = pd.read_parquet("../data/processed/avaliacoes.parquet")
```

## Funcão para classificar sentimentos

```python
def classificar_sentimento(score):
    if score >= 0.75: return 'Muito Positivo'
    elif score >= 0.25: return 'Positivo'
    elif score > -0.25: return 'Neutro'
    elif score >= -0.75: return 'Negativo'
    else: return 'Muito Negativo'
```

## Vader

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
```

```python
vader = SentimentIntensityAnalyzer()
avaliacoes['vader_score'] = avaliacoes['review_comment'].apply(lambda x: vader.polarity_scores(str(x))['compound'])
avaliacoes['vader_label'] = avaliacoes['vader_score'].apply(classificar_sentimento)
display(avaliacoes.head(10))
```

## LeIA

```python
from LeIA import SentimentIntensityAnalyzer
```

```python
leia = SentimentIntensityAnalyzer()
avaliacoes['leia_score'] = avaliacoes['review_comment'].apply(lambda x: leia.polarity_scores(str(x))['compound'])
avaliacoes['leia_label'] = avaliacoes['leia_score'].apply(classificar_sentimento)
display(avaliacoes.head(10))
```

## TextBlob

```python
from textblob import TextBlob
```

```python
avaliacoes['textblob_score'] = avaliacoes['review_comment'].apply(lambda x: TextBlob(str(x)).sentiment.polarity)
avaliacoes['textblob_label'] = avaliacoes['textblob_score'].apply(classificar_sentimento)
display(avaliacoes.head(10))
```
