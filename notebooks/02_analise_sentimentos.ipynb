# **Análise de Sentimentos**

## **Importação das bibliotecas e leitura dos dados**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
avaliacoes = pd.read_parquet("../data/processed/avaliacoes.parquet")
```

## **Função para classificar sentimentos**

Essa função classifica os sentimentos com base nos scores de polaridade gerados pelos modelos, atribuindo rótulos como "Muito Positivo", "Positivo", "Neutro", "Negativo" ou "Muito Negativo".

```python
def classificar_sentimento(score):
    if score >= 0.75: return 'Muito Positivo'
    elif score >= 0.25: return 'Positivo'
    elif score > -0.25: return 'Neutro'
    elif score >= -0.75: return 'Negativo'
    else: return 'Muito Negativo'
```

## **Aplicação dos modelos de análise de sentimentos**

Nessa seção, aplicamos os modelos Vader, leIA e TextBlob para analisar os sentimentos expressos nos comentários da base de dados. Todos os modelos devolvem scores de polaridade, ou seja, valores numéricos que indicam a intensidade do sentimento presente em cada comentário. Para todos, o intervalo de polaridade varia de -1 a 1, onde -1 representa um sentimento extremamente negativo, 0 representa um sentimento neutro e 1 representa um sentimento extremamente positivo. Após obter os scores de polaridade, utilizamos a função de classificação para categorizar os sentimentos em rótulos compreensíveis, facilitando a interpretação dos resultados.

### **Vader**

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
```

```python
vader = SentimentIntensityAnalyzer()
avaliacoes['vader_score'] = avaliacoes['review_comment'].apply(lambda x: vader.polarity_scores(str(x))['compound'])
avaliacoes['vader_label'] = avaliacoes['vader_score'].apply(classificar_sentimento)
display(avaliacoes.head(10))
```

### **LeIA**

```python
from leia import SentimentIntensityAnalyzer
```

```python
leia = SentimentIntensityAnalyzer()
avaliacoes['leia_score'] = avaliacoes['review_comment'].apply(lambda x: leia.polarity_scores(str(x))['compound'])
avaliacoes['leia_label'] = avaliacoes['leia_score'].apply(classificar_sentimento)
display(avaliacoes.head(10))
```

### **TextBlob**

```python
from textblob import TextBlob
```

```python
avaliacoes['textblob_score'] = avaliacoes['review_comment'].apply(lambda x: TextBlob(str(x)).sentiment.polarity)
avaliacoes['textblob_label'] = avaliacoes['textblob_score'].apply(classificar_sentimento)
display(avaliacoes.head(10))
```
